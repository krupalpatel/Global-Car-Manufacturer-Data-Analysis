```python
# =====================================================
# 1. IMPORT LIBRARIES
# =====================================================
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import plotly.graph_objects as go
import plotly.express as px
import plotly.io as pio

pio.renderers.default = 'iframe'

from sklearn.preprocessing import MinMaxScaler
from sklearn.cluster import KMeans

plt.style.use('dark_background')
```

```python
# =====================================================
# 2. LOAD DATA
# =====================================================
file_path = "/kaggle/input/datasets/krupalpatel07/toyota-motor-corporation/Toyota Motor Corporation.csv"
df = pd.read_csv(file_path)
```

```python
# =====================================================
# 3. PREPROCESSING
# =====================================================
df.columns = [c.lower() for c in df.columns]

df['date'] = pd.to_datetime(df['date'])
df = df.sort_values('date')

df.set_index('date', inplace=True)
```

```python
# =====================================================
# 4. CYBER HEADER
# =====================================================
from IPython.display import display, HTML

def cyber_header(title):
    display(HTML(f"""
    <div style="
        background: linear-gradient(135deg,#ff0080,#7928ca,#00d4ff);
        padding:22px;
        border-radius:18px;
        margin-top:22px;
        box-shadow:0px 0px 25px rgba(255,0,128,0.6);
        backdrop-filter: blur(10px);
    ">
        <h1 style="
            color:white;
            text-align:center;
            font-size:38px;
            letter-spacing:2px;">
            {title}
        </h1>
    </div>
    """))

cyber_header("📊 Toyota Market Ecosystem")
```

```python
# =====================================================
# 5. PRICE VISUALIZATION
# =====================================================
fig = go.Figure()

fig.add_trace(go.Scatter(
    x=df.index,
    y=df['close'],
    mode='lines',
    line=dict(color='#00d4ff', width=2),
    name='Close Price'
))

fig.update_layout(
    template='plotly_dark',
    title='Toyota Price Evolution',
    height=600
)

fig.show()
```

```python
# =====================================================
# 6. FEATURE ENGINEERING
# =====================================================
cyber_header("🧠 Adaptive Feature Matrix")

df['returns'] = df['close'].pct_change()

df['volatility'] = df['returns'].rolling(20).std()

df['momentum'] = df['close'].pct_change(10)

df['volume_force'] = (
    (df['close'] - df['open']) * df['volume']
)

df['trend_strength'] = (
    df['close'].rolling(20).mean() -
    df['close'].rolling(50).mean()
)

df = df.dropna()
```

```python
# =====================================================
# 7. MARKET STATE CLUSTERING
# =====================================================
cyber_header("⚡ Market State Clustering")

features = df[
    ['returns','volatility','momentum','trend_strength']
]

scaler = MinMaxScaler()

scaled = scaler.fit_transform(features)

kmeans = KMeans(
    n_clusters=4,
    random_state=42
)

df['state'] = kmeans.fit_predict(scaled)

fig = px.scatter(
    df,
    x=df.index,
    y='close',
    color=df['state'].astype(str),
    title='Market State Regimes'
)

fig.show()
```

```python
# =====================================================
# 8. REINFORCEMENT-STYLE REWARD ENGINE
# =====================================================
cyber_header("🎮 Reinforcement Reward System")

# reward proxy
df['reward'] = np.where(
    df['returns'] > 0,
    1,
    -1
)

reward_curve = df['reward'].cumsum()

fig = px.line(
    reward_curve,
    title='Cumulative Reward Curve'
)

fig.show()
```

```python
# =====================================================
# 9. ADAPTIVE SIGNAL SYSTEM
# =====================================================
cyber_header("🎯 Adaptive Alpha Signals")

signal = (
    (df['momentum'] > 0) &
    (df['trend_strength'] > 0) &
    (df['volatility'] < df['volatility'].rolling(50).mean())
)

df['signal'] = signal.astype(int)

fig = go.Figure()

fig.add_trace(go.Scatter(
    x=df.index,
    y=df['close'],
    name='Price'
))

fig.add_trace(go.Scatter(
    x=df.index[df['signal']==1],
    y=df['close'][df['signal']==1],
    mode='markers',
    marker=dict(
        color='#ff0080',
        size=8
    ),
    name='Adaptive Buy'
))

fig.update_layout(
    template='plotly_dark',
    title='Adaptive Trading Engine'
)

fig.show()
```

```python
# =====================================================
# 10. MARKET ENERGY MAP
# =====================================================
cyber_header("🌌 Market Energy Map")

df['energy'] = (
    np.abs(df['returns']) *
    df['volume']
)

fig = px.area(
    df,
    y='energy',
    title='Market Energy Flow'
)

fig.show()
```

```python
# =====================================================
# 11. CYBER VOLATILITY MATRIX
# =====================================================
cyber_header("🔥 Cyber Volatility Matrix")

vol_matrix = df['returns'].rolling(30).std()

fig = px.imshow(
    vol_matrix.to_frame().T,
    aspect='auto',
    title='Volatility Matrix'
)

fig.show()
```

```python
# =====================================================
# 12. STRATEGY PERFORMANCE
# =====================================================
cyber_header("📈 Strategy Performance")

df['strategy_returns'] = (
    df['signal'].shift(1) *
    df['returns']
)

df['cum_market'] = (
    1 + df['returns']
).cumprod()

df['cum_strategy'] = (
    1 + df['strategy_returns']
).cumprod()

fig = go.Figure()

fig.add_trace(go.Scatter(
    x=df.index,
    y=df['cum_market'],
    name='Market'
))

fig.add_trace(go.Scatter(
    x=df.index,
    y=df['cum_strategy'],
    name='Adaptive Strategy'
))

fig.update_layout(
    template='plotly_dark',
    title='Strategy vs Market'
)

fig.show()
```

```python
# =====================================================
# 13. FINAL INSIGHTS
# =====================================================
cyber_header("📌 Nexus Intelligence")

print("""
1. Markets evolve through different hidden states.
2. Adaptive systems outperform static logic.
3. Reinforcement-style rewards improve decision intuition.
4. Market energy reveals participation intensity.
5. Multi-factor adaptive models create resilient alpha.
""")

# =====================================================
# END
# =====================================================
```
